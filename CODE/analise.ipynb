```python
# ============================================================
# CONFIGURAÇÃO
# ============================================================

CAMINHO_EXCEL = Path(
    r'C:\Users\thiago.deoliveira\Desktop\VARIACAO_NOTA\DADOS\dados_avaliacao.xlsx'
)

OUTPUT_RESULTADO = Path(
    r'C:\Users\thiago.deoliveira\Desktop\VARIACAO_NOTA\DADOS\resultado_variacao.xlsx'
)


# ============================================================
# NOTAS VÁLIDAS
# ============================================================

NOTAS_VALIDAS = {
    'Bom',
    'Muito Bom',
    'Ruim',
    'Muito Ruim'
}


# ============================================================
# COLUNAS DE NOTA ENCONTRADAS NA SUA PLANILHA
# ============================================================

COLS_AVALIACAO = [
    'a1_cons',
    'a1_limp',
    'a1_mob',

    'a2_cons',
    'a2_mob',
    'a2_limp',

    'a3bf_cons',
    'a3bf_repo',
    'a3bf_limp',

    'a3bm_limp',
    'a3bm_cons',
    'a3bm_repo',

    'a4_cons',

    'a5_cons',
    'a5_limp',

    'a6coz_equip',
    'a6coz_limp',
    'a6coz_armz',

    'a6ref_armz',
    'a6ref_cons',
    'a6ref_mob',
    'a6ref_limp',

    'a6_quali',

    'a7_jardinag',
    'a7_cons',
    'a7_limp_ext',
    'a7_acesso',
    'a7_dedet',
    'a7_pint'
]


# ============================================================
# LEITURA
# ============================================================

print('Lendo planilha...')

df = pd.read_excel(
    CAMINHO_EXCEL,
    sheet_name='result'
)


# ============================================================
# PADRONIZAÇÃO DOS NOMES
# ============================================================

df = df.rename(
    columns={
        'nm_escola': 'escola',
        'nomecompleto': 'avaliador'
    }
)


# ============================================================
# VALIDAÇÃO
# ============================================================

obrigatorias = [
    'escola',
    'avaliador',
    'dt_inicio'
]

faltantes = [
    c for c in obrigatorias
    if c not in df.columns
]

if faltantes:
    raise ValueError(
        f'Colunas obrigatórias ausentes: {faltantes}'
    )


# verifica quais colunas de avaliação existem
cols_avaliacao = [
    c for c in COLS_AVALIACAO
    if c in df.columns
]


print(f'Registros: {len(df):,}')
print(f'Critérios de avaliação: {len(cols_avaliacao)}')


# ============================================================
# FORMATO LONGO
#
# Cada linha passa a representar:
#
# escola
# avaliador
# data
# criterio
# nota
#
# ============================================================

long_df = df[
    [
        'escola',
        'avaliador',
        'dt_inicio'
    ]
    + cols_avaliacao
].melt(

    id_vars=[
        'escola',
        'avaliador',
        'dt_inicio'
    ],

    value_vars=cols_avaliacao,

    var_name='criterio',

    value_name='nota'
)


# ============================================================
# LIMPEZA
# ============================================================

long_df = long_df.dropna(
    subset=['nota']
).copy()


long_df['nota'] = (
    long_df['nota']
    .astype(str)
    .str.strip()
)


# mantém EXCLUSIVAMENTE as quatro notas
long_df = long_df[
    long_df['nota'].isin(NOTAS_VALIDAS)
].copy()


# data
long_df['dt_inicio'] = pd.to_datetime(
    long_df['dt_inicio'],
    errors='coerce'
)


# ordena cronologicamente
long_df = long_df.sort_values(
    [
        'escola',
        'avaliador',
        'criterio',
        'dt_inicio'
    ]
)


# ============================================================
# FUNÇÕES DE AGREGAÇÃO
# ============================================================

def sequencia_notas(serie):
    """
    Mostra exatamente a sequência das notas
    encontradas na planilha.
    """
    return ' -> '.join(
        serie.astype(str)
    )


def sequencia_datas(serie):
    """
    Mostra as datas correspondentes às avaliações.
    """
    return ' -> '.join(
        serie.dt.strftime('%d/%m/%Y').fillna('')
    )


def categorias_encontradas(serie):
    """
    Lista somente as categorias que realmente apareceram.
    Sem ordenar e sem atribuir peso.
    """
    return ' | '.join(
        pd.unique(serie.astype(str))
    )


# ============================================================
# AGRUPAMENTO PRINCIPAL
#
# MESMA ESCOLA
# + MESMO AVALIADOR
# + MESMO CRITÉRIO
# ============================================================

resultado = (
    long_df
    .groupby(
        [
            'escola',
            'avaliador',
            'criterio'
        ],
        dropna=False,
        sort=False
    )
    .agg(

        qtd_avaliacoes=(
            'nota',
            'size'
        ),

        qtd_notas_diferentes=(
            'nota',
            'nunique'
        ),

        notas_encontradas=(
            'nota',
            categorias_encontradas
        ),

        sequencia_notas=(
            'nota',
            sequencia_notas
        ),

        sequencia_datas=(
            'dt_inicio',
            sequencia_datas
        ),

        qtd_muito_bom=(
            'nota',
            lambda x: (x == 'Muito Bom').sum()
        ),

        qtd_bom=(
            'nota',
            lambda x: (x == 'Bom').sum()
        ),

        qtd_ruim=(
            'nota',
            lambda x: (x == 'Ruim').sum()
        ),

        qtd_muito_ruim=(
            'nota',
            lambda x: (x == 'Muito Ruim').sum()
        )
    )
    .reset_index()
)


# ============================================================
# SÓ É POSSÍVEL MEDIR VARIAÇÃO SE EXISTIREM
# PELO MENOS DUAS AVALIAÇÕES
# ============================================================

resultado = resultado[
    resultado['qtd_avaliacoes'] >= 2
].copy()


# ============================================================
# HOUVE VARIAÇÃO?
#
# Nenhuma pontuação ou peso é criado.
#
# 1 categoria diferente  -> não variou
# mais de 1 categoria    -> variou
# ============================================================

resultado['houve_variacao'] = resultado[
    'qtd_notas_diferentes'
].gt(1).map({
    True: 'Sim',
    False: 'Não'
})


# ============================================================
# ORGANIZAÇÃO DAS COLUNAS
# ============================================================

resultado = resultado[
    [
        'escola',
        'avaliador',
        'criterio',

        'qtd_avaliacoes',
        'qtd_notas_diferentes',

        'houve_variacao',

        'notas_encontradas',
        'sequencia_notas',
        'sequencia_datas',

        'qtd_muito_bom',
        'qtd_bom',
        'qtd_ruim',
        'qtd_muito_ruim'
    ]
]


# ============================================================
# PRIORIZA NA VISUALIZAÇÃO OS CASOS QUE VARIARAM
# ============================================================

resultado = resultado.sort_values(
    [
        'houve_variacao',
        'qtd_notas_diferentes',
        'qtd_avaliacoes'
    ],
    ascending=[
        False,
        False,
        False
    ]
)


# ============================================================
# RESUMO:
# ESCOLA + AVALIADOR
# ============================================================

resumo_escola_avaliador = (
    resultado
    .groupby(
        [
            'escola',
            'avaliador'
        ],
        as_index=False
    )
    .agg(

        qtd_criterios_comparados=(
            'criterio',
            'count'
        ),

        qtd_criterios_com_variacao=(
            'houve_variacao',
            lambda x: (x == 'Sim').sum()
        ),

        qtd_criterios_sem_variacao=(
            'houve_variacao',
            lambda x: (x == 'Não').sum()
        )
    )
)


# percentual puramente descritivo
resumo_escola_avaliador[
    'percentual_criterios_com_variacao'
] = (

    resumo_escola_avaliador[
        'qtd_criterios_com_variacao'
    ]

    /

    resumo_escola_avaliador[
        'qtd_criterios_comparados'
    ]

    * 100
).round(2)


# ============================================================
# RESUMO POR CRITÉRIO / AMBIENTE
# ============================================================

resumo_criterio = (
    resultado
    .groupby(
        'criterio',
        as_index=False
    )
    .agg(

        qtd_comparacoes=(
            'criterio',
            'size'
        ),

        qtd_com_variacao=(
            'houve_variacao',
            lambda x: (x == 'Sim').sum()
        ),

        qtd_sem_variacao=(
            'houve_variacao',
            lambda x: (x == 'Não').sum()
        )
    )
)


resumo_criterio[
    'percentual_com_variacao'
] = (

    resumo_criterio[
        'qtd_com_variacao'
    ]

    /

    resumo_criterio[
        'qtd_comparacoes'
    ]

    * 100
).round(2)


resumo_criterio = resumo_criterio.sort_values(
    'percentual_com_variacao',
    ascending=False
)


# ============================================================
# REGISTROS ORIGINAIS QUE ENTRARAM NA COMPARAÇÃO
# ============================================================

detalhe = long_df.copy()


# ============================================================
# EXPORTAÇÃO
# ============================================================

print('\nExportando resultado...')


with pd.ExcelWriter(
    OUTPUT_RESULTADO,
    engine='openpyxl'
) as writer:

    # --------------------------------------------------------
    # comparação principal
    # --------------------------------------------------------

    resultado.to_excel(
        writer,
        sheet_name='variacao_detalhada',
        index=False
    )

    # --------------------------------------------------------
    # resumo escola + avaliador
    # --------------------------------------------------------

    resumo_escola_avaliador.to_excel(
        writer,
        sheet_name='escola_avaliador',
        index=False
    )

    # --------------------------------------------------------
    # resumo por critério
    # --------------------------------------------------------

    resumo_criterio.to_excel(
        writer,
        sheet_name='criterios',
        index=False
    )

    # --------------------------------------------------------
    # dados que efetivamente participaram
    # --------------------------------------------------------

    detalhe.to_excel(
        writer,
        sheet_name='dados_comparados',
        index=False
    )


# ============================================================
# RESULTADO NO TERMINAL
# ============================================================

print('\n')
print('=' * 100)
print('VARIAÇÃO - MESMA ESCOLA + MESMO AVALIADOR + MESMO CRITÉRIO')
print('=' * 100)


print(
    resultado[
        resultado['houve_variacao'] == 'Sim'
    ]
    .head(30)
    .to_string(
        index=False,
        max_colwidth=50
    )
)


print('\n')
print('=' * 100)
print('RESUMO POR ESCOLA + AVALIADOR')
print('=' * 100)


print(
    resumo_escola_avaliador
    .sort_values(
        'percentual_criterios_com_variacao',
        ascending=False
    )
    .head(30)
    .to_string(index=False)
)


print('\n')
print('=' * 100)
print('RESUMO POR CRITÉRIO')
print('=' * 100)


print(
    resumo_criterio
    .head(30)
    .to_string(index=False)
)


print(
    f'\nArquivo salvo em:\n{OUTPUT_RESULTADO}'
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from openpyxl import load_workbook
from openpyxl.styles import Font, Alignment
from openpyxl.utils import get_column_letter


# ============================================================
# CONFIGURAÇÃO
# ============================================================

CAMINHO_EXCEL = Path(
    r'C:\Users\thiago.deoliveira\Desktop\VARIACAO_NOTA\DADOS\dados_avaliacao.xlsx'
)

PASTA_SAIDA = Path(
    r'C:\Users\thiago.deoliveira\Desktop\VARIACAO_NOTA\RESULTADOS_EDA'
)

PASTA_GRAFICOS = PASTA_SAIDA / 'graficos'

ARQUIVO_RESULTADO = (
    PASTA_SAIDA /
    'analise_exploratoria_variacao.xlsx'
)


PASTA_SAIDA.mkdir(
    parents=True,
    exist_ok=True
)

PASTA_GRAFICOS.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# CATEGORIAS DE NOTA
#
# IMPORTANTE:
# Não existe conversão para 1, 2, 3 ou 4.
#
# As categorias permanecem exatamente como estão na planilha.
# ============================================================

NOTAS = [
    'Muito Bom',
    'Bom',
    'Ruim',
    'Muito Ruim'
]


# ============================================================
# CRITÉRIOS DE AVALIAÇÃO
#
# Somente colunas que realmente possuem:
#
# Muito Bom
# Bom
# Ruim
# Muito Ruim
# ============================================================

COLS_AVALIACAO = [

    # A1
    'a1_cons',
    'a1_limp',
    'a1_mob',

    # A2
    'a2_cons',
    'a2_mob',
    'a2_limp',

    # A3 banheiro feminino
    'a3bf_cons',
    'a3bf_repo',
    'a3bf_limp',

    # A3 banheiro masculino
    'a3bm_limp',
    'a3bm_cons',
    'a3bm_repo',

    # A4
    'a4_cons',

    # A5
    'a5_cons',
    'a5_limp',

    # A6 cozinha
    'a6coz_equip',
    'a6coz_limp',
    'a6coz_armz',

    # A6 refeitório
    'a6ref_armz',
    'a6ref_cons',
    'a6ref_mob',
    'a6ref_limp',

    # A6
    'a6_quali',

    # A7
    'a7_jardinag',
    'a7_cons',
    'a7_limp_ext',
    'a7_acesso',
    'a7_dedet',
    'a7_pint'
]


# ============================================================
# FUNÇÃO PARA IDENTIFICAR O BLOCO
# ============================================================

def identificar_bloco(criterio):

    if criterio.startswith('a1_'):
        return 'A1'

    if criterio.startswith('a2_'):
        return 'A2'

    if criterio.startswith('a3bf_'):
        return 'A3BF'

    if criterio.startswith('a3bm_'):
        return 'A3BM'

    if criterio.startswith('a4_'):
        return 'A4'

    if criterio.startswith('a5_'):
        return 'A5'

    if criterio.startswith('a6coz_'):
        return 'A6COZ'

    if criterio.startswith('a6ref_'):
        return 'A6REF'

    if criterio.startswith('a6_'):
        return 'A6'

    if criterio.startswith('a7_'):
        return 'A7'

    return 'OUTRO'


# ============================================================
# LEITURA
# ============================================================

print('=' * 100)
print('ANÁLISE EXPLORATÓRIA DAS AVALIAÇÕES')
print('=' * 100)

print('\n1. Lendo planilha...')

df = pd.read_excel(
    CAMINHO_EXCEL,
    sheet_name='result'
)


print(f'Registros encontrados: {len(df):,}')
print(f'Colunas encontradas:   {len(df.columns):,}')


# ============================================================
# PADRONIZAÇÃO DOS NOMES PRINCIPAIS
# ============================================================

df = df.rename(
    columns={
        'nm_escola': 'escola',
        'nomecompleto': 'avaliador'
    }
)


# ============================================================
# VALIDAÇÕES
# ============================================================

COLUNAS_OBRIGATORIAS = [
    'id_registro',
    'dt_inicio',
    'escola',
    'avaliador'
]


faltantes = [
    c
    for c in COLUNAS_OBRIGATORIAS
    if c not in df.columns
]


if faltantes:

    raise ValueError(
        f'Colunas obrigatórias não encontradas: {faltantes}'
    )


# ============================================================
# DATA
# ============================================================

df['dt_inicio'] = pd.to_datetime(
    df['dt_inicio'],
    errors='coerce'
)


# ============================================================
# CRITÉRIOS REALMENTE EXISTENTES
# ============================================================

criterios_existentes = [
    c
    for c in COLS_AVALIACAO
    if c in df.columns
]


print(
    f'Critérios categóricos utilizados: '
    f'{len(criterios_existentes)}'
)


# ============================================================
# COLUNAS AUXILIARES DE AMBIENTE
# ============================================================

COLUNAS_AMBIENTE = [
    'a1_salaAvaliada',
    'a2_ambienteAvaliado',
    'a4_armz'
]


for coluna in COLUNAS_AMBIENTE:

    if coluna not in df.columns:
        df[coluna] = pd.NA


# ============================================================
# TRANSFORMAÇÃO WIDE -> LONG
# ============================================================

print('\n2. Transformando dados...')


id_vars = [
    'id_registro',
    'dt_inicio',
    'escola',
    'avaliador',
    'a1_salaAvaliada',
    'a2_ambienteAvaliado',
    'a4_armz'
]


long_df = df[
    id_vars + criterios_existentes
].melt(

    id_vars=id_vars,

    value_vars=criterios_existentes,

    var_name='criterio',

    value_name='nota'
)


# ============================================================
# LIMPEZA DAS NOTAS
# ============================================================

long_df['nota'] = (
    long_df['nota']
    .astype('string')
    .str.strip()
)


# mantém APENAS as quatro categorias reais
long_df = long_df[
    long_df['nota'].isin(NOTAS)
].copy()


print(
    f'Avaliações categóricas válidas: '
    f'{len(long_df):,}'
)


# ============================================================
# BLOCO
# ============================================================

long_df['bloco'] = (
    long_df['criterio']
    .map(identificar_bloco)
)


# ============================================================
# AMBIENTE
#
# A análise respeita o ambiente real quando a planilha
# possui uma coluna específica para identificá-lo.
#
# A1 -> a1_salaAvaliada
# A2 -> a2_ambienteAvaliado
# A4 -> a4_armz
#
# Para os demais, o próprio bloco identifica o ambiente.
# ============================================================

long_df['ambiente'] = long_df['bloco']


# ------------------------------------------------------------
# A1
# ------------------------------------------------------------

mask = (
    long_df['bloco'] == 'A1'
)

long_df.loc[
    mask,
    'ambiente'
] = (
    long_df.loc[
        mask,
        'a1_salaAvaliada'
    ]
    .astype('string')
    .str.strip()
)


# ------------------------------------------------------------
# A2
# ------------------------------------------------------------

mask = (
    long_df['bloco'] == 'A2'
)

long_df.loc[
    mask,
    'ambiente'
] = (
    long_df.loc[
        mask,
        'a2_ambienteAvaliado'
    ]
    .astype('string')
    .str.strip()
)


# ------------------------------------------------------------
# A4
# ------------------------------------------------------------

mask = (
    long_df['bloco'] == 'A4'
)

long_df.loc[
    mask,
    'ambiente'
] = (
    long_df.loc[
        mask,
        'a4_armz'
    ]
    .astype('string')
    .str.strip()
)


# ============================================================
# MÊS
# ============================================================

long_df['mes'] = (
    long_df['dt_inicio']
    .dt.to_period('M')
    .astype(str)
)


# ============================================================
# 1 - RESUMO GERAL
# ============================================================

print('\n3. Calculando indicadores gerais...')


total_registros = len(df)

total_escolas = (
    df['escola']
    .nunique()
)

total_avaliadores = (
    df['avaliador']
    .nunique()
)

total_avaliacoes = len(long_df)


data_inicio = (
    df['dt_inicio']
    .min()
)

data_fim = (
    df['dt_inicio']
    .max()
)


resumo_geral = pd.DataFrame({

    'indicador': [

        'Registros',
        'Escolas',
        'Avaliadores',
        'Avaliações categóricas',
        'Data inicial',
        'Data final'

    ],

    'valor': [

        total_registros,
        total_escolas,
        total_avaliadores,
        total_avaliacoes,
        data_inicio,
        data_fim

    ]
})


# ============================================================
# 2 - DISTRIBUIÇÃO GERAL DAS NOTAS
# ============================================================

distribuicao_notas = (

    long_df['nota']
    .value_counts()
    .reindex(
        NOTAS,
        fill_value=0
    )
    .rename_axis('nota')
    .reset_index(
        name='quantidade'
    )
)


distribuicao_notas['percentual'] = (

    distribuicao_notas['quantidade']

    /

    distribuicao_notas[
        'quantidade'
    ].sum()

    * 100
)


distribuicao_notas[
    'percentual'
] = (

    distribuicao_notas[
        'percentual'
    ]
    .round(2)
)


# ============================================================
# BASE PARA COMPARAÇÃO DE VARIAÇÃO
#
# Para comparar avaliações é necessário conhecer:
#
# escola
# avaliador
# critério
# ambiente
#
# Linhas sem identificação do ambiente nos blocos que
# dependem dessa informação não entram na comparação.
#
# Elas continuam participando da distribuição geral.
# ============================================================

base_comparacao = long_df[

    long_df['escola'].notna()

    &

    long_df['avaliador'].notna()

    &

    long_df['ambiente'].notna()

].copy()


base_comparacao = base_comparacao[

    base_comparacao[
        'ambiente'
    ]
    .astype('string')
    .str.strip()
    .ne('')

].copy()


# ============================================================
# CHAVE DE COMPARAÇÃO
#
# MESMA ESCOLA
# MESMO AVALIADOR
# MESMO CRITÉRIO
# MESMO AMBIENTE
# ============================================================

CHAVES = [

    'escola',
    'avaliador',
    'criterio',
    'ambiente'

]


# ============================================================
# 3 - VARIAÇÃO DETALHADA
# ============================================================

print('\n4. Calculando variação categórica...')


grupo = (

    base_comparacao

    .groupby(
        CHAVES,
        dropna=False,
        sort=False
    )

    .agg(

        qtd_avaliacoes=(
            'nota',
            'size'
        ),

        qtd_notas_diferentes=(
            'nota',
            'nunique'
        ),

        primeira_avaliacao=(
            'dt_inicio',
            'min'
        ),

        ultima_avaliacao=(
            'dt_inicio',
            'max'
        )

    )

    .reset_index()

)


# ============================================================
# CONTAGEM DE CADA CATEGORIA POR GRUPO
# ============================================================

contagem_notas = (

    base_comparacao

    .pivot_table(

        index=CHAVES,

        columns='nota',

        values='id_registro',

        aggfunc='count',

        fill_value=0

    )

    .reindex(
        columns=NOTAS,
        fill_value=0
    )

    .reset_index()

)


contagem_notas = contagem_notas.rename(

    columns={

        'Muito Bom':
            'qtd_muito_bom',

        'Bom':
            'qtd_bom',

        'Ruim':
            'qtd_ruim',

        'Muito Ruim':
            'qtd_muito_ruim'

    }

)


variacao_detalhada = grupo.merge(

    contagem_notas,

    on=CHAVES,

    how='left'

)


# ============================================================
# SOMENTE GRUPOS COM PELO MENOS DUAS AVALIAÇÕES
# ============================================================

variacao_detalhada = variacao_detalhada[

    variacao_detalhada[
        'qtd_avaliacoes'
    ] >= 2

].copy()


# ============================================================
# HOUVE VARIAÇÃO?
#
# Se apareceu mais de uma categoria:
# Sim
#
# Caso contrário:
# Não
#
# Nenhum score numérico é criado.
# ============================================================

variacao_detalhada[
    'houve_variacao'
] = np.where(

    variacao_detalhada[
        'qtd_notas_diferentes'
    ] > 1,

    'Sim',

    'Não'

)


# ============================================================
# PERCENTUAIS INTERNOS
# ============================================================

for coluna in [

    'qtd_muito_bom',
    'qtd_bom',
    'qtd_ruim',
    'qtd_muito_ruim'

]:

    nome_pct = coluna.replace(
        'qtd_',
        'pct_'
    )

    variacao_detalhada[nome_pct] = (

        variacao_detalhada[coluna]

        /

        variacao_detalhada[
            'qtd_avaliacoes'
        ]

        * 100

    ).round(2)


# ============================================================
# ORDENAÇÃO
# ============================================================

variacao_detalhada = (

    variacao_detalhada

    .sort_values(

        [
            'houve_variacao',
            'qtd_notas_diferentes',
            'qtd_avaliacoes'
        ],

        ascending=[
            False,
            False,
            False
        ]

    )

)


# ============================================================
# 4 - TRANSIÇÕES ENTRE VISITAS
#
# Aqui analisamos:
#
# Bom -> Bom
# Bom -> Muito Bom
# Bom -> Ruim
# Ruim -> Bom
# etc.
#
# Não existe conversão numérica.
# ============================================================

print('\n5. Calculando transições entre visitas...')


base_transicoes = (

    base_comparacao

    .sort_values(

        CHAVES
        +
        [
            'dt_inicio',
            'id_registro'
        ]

    )

    .copy()

)


base_transicoes[
    'nota_anterior'
] = (

    base_transicoes

    .groupby(
        CHAVES,
        dropna=False
    )['nota']

    .shift(1)

)


base_transicoes[
    'data_anterior'
] = (

    base_transicoes

    .groupby(
        CHAVES,
        dropna=False
    )['dt_inicio']

    .shift(1)

)


transicoes = base_transicoes[

    base_transicoes[
        'nota_anterior'
    ].notna()

].copy()


transicoes[
    'mudou_nota'
] = np.where(

    transicoes['nota']
    ==
    transicoes['nota_anterior'],

    'Não',

    'Sim'

)


transicoes[
    'transicao'
] = (

    transicoes[
        'nota_anterior'
    ].astype(str)

    +

    ' -> '

    +

    transicoes[
        'nota'
    ].astype(str)

)


# ============================================================
# RESUMO DAS TRANSIÇÕES
# ============================================================

resumo_transicoes = (

    transicoes

    .groupby(
        'transicao',
        as_index=False
    )

    .size()

    .rename(
        columns={
            'size':
                'quantidade'
        }
    )

)


resumo_transicoes[
    'percentual'
] = (

    resumo_transicoes[
        'quantidade'
    ]

    /

    resumo_transicoes[
        'quantidade'
    ].sum()

    * 100

).round(2)


resumo_transicoes = (

    resumo_transicoes

    .sort_values(
        'quantidade',
        ascending=False
    )

)


# ============================================================
# RESUMO ESTABILIDADE ENTRE VISITAS
# ============================================================

estabilidade_visitas = (

    transicoes[
        'mudou_nota'
    ]

    .value_counts()

    .rename_axis(
        'mudou_nota'
    )

    .reset_index(
        name='quantidade'
    )

)


estabilidade_visitas[
    'percentual'
] = (

    estabilidade_visitas[
        'quantidade'
    ]

    /

    estabilidade_visitas[
        'quantidade'
    ].sum()

    * 100

).round(2)


# ============================================================
# FUNÇÃO DE RESUMO
# ============================================================

def gerar_resumo_variacao(
    coluna_grupo
):

    resumo = (

        variacao_detalhada

        .groupby(
            coluna_grupo,
            as_index=False,
            dropna=False
        )

        .agg(

            qtd_grupos_comparados=(
                'houve_variacao',
                'size'
            ),

            qtd_grupos_com_variacao=(
                'houve_variacao',
                lambda x:
                (x == 'Sim').sum()
            ),

            qtd_grupos_sem_variacao=(
                'houve_variacao',
                lambda x:
                (x == 'Não').sum()
            ),

            qtd_avaliacoes=(
                'qtd_avaliacoes',
                'sum'
            )

        )

    )


    resumo[
        'pct_grupos_com_variacao'
    ] = (

        resumo[
            'qtd_grupos_com_variacao'
        ]

        /

        resumo[
            'qtd_grupos_comparados'
        ]

        * 100

    ).round(2)


    return resumo


# ============================================================
# 5 - VARIAÇÃO POR CRITÉRIO
# ============================================================

variacao_criterio = (
    gerar_resumo_variacao(
        'criterio'
    )
)


# ============================================================
# DISTRIBUIÇÃO DAS NOTAS POR CRITÉRIO
# ============================================================

notas_criterio = (

    long_df

    .groupby(
        [
            'criterio',
            'nota'
        ]
    )

    .size()

    .unstack(
        fill_value=0
    )

    .reindex(
        columns=NOTAS,
        fill_value=0
    )

    .reset_index()

)


notas_criterio[
    'total_notas'
] = (

    notas_criterio[
        NOTAS
    ]
    .sum(axis=1)

)


for nota in NOTAS:

    nome = (
        'pct_'
        +
        nota.lower()
        .replace(' ', '_')
    )

    notas_criterio[nome] = (

        notas_criterio[nota]

        /

        notas_criterio[
            'total_notas'
        ]

        * 100

    ).round(2)


# ============================================================
# RUIM + MUITO RUIM
# ============================================================

notas_criterio[
    'pct_ruim_muito_ruim'
] = (

    (
        notas_criterio['Ruim']
        +
        notas_criterio['Muito Ruim']
    )

    /

    notas_criterio[
        'total_notas'
    ]

    * 100

).round(2)


variacao_criterio = (

    variacao_criterio

    .merge(

        notas_criterio,

        on='criterio',

        how='left'

    )

)


variacao_criterio = (

    variacao_criterio

    .sort_values(

        'pct_grupos_com_variacao',

        ascending=False

    )

)


# ============================================================
# 6 - VARIAÇÃO POR BLOCO
# ============================================================

criterio_bloco = (

    long_df[
        [
            'criterio',
            'bloco'
        ]
    ]

    .drop_duplicates()

)


variacao_bloco = (

    variacao_detalhada

    .merge(

        criterio_bloco,

        on='criterio',

        how='left'

    )

)


variacao_bloco = (

    variacao_bloco

    .groupby(
        'bloco',
        as_index=False
    )

    .agg(

        qtd_grupos_comparados=(
            'houve_variacao',
            'size'
        ),

        qtd_grupos_com_variacao=(
            'houve_variacao',
            lambda x:
            (x == 'Sim').sum()
        ),

        qtd_avaliacoes=(
            'qtd_avaliacoes',
            'sum'
        )

    )

)


variacao_bloco[
    'pct_grupos_com_variacao'
] = (

    variacao_bloco[
        'qtd_grupos_com_variacao'
    ]

    /

    variacao_bloco[
        'qtd_grupos_comparados'
    ]

    * 100

).round(2)


# ============================================================
# NOTAS POR BLOCO
# ============================================================

notas_bloco = (

    long_df

    .groupby(
        [
            'bloco',
            'nota'
        ]
    )

    .size()

    .unstack(
        fill_value=0
    )

    .reindex(
        columns=NOTAS,
        fill_value=0
    )

    .reset_index()

)


notas_bloco[
    'total_notas'
] = (

    notas_bloco[
        NOTAS
    ].sum(axis=1)

)


notas_bloco[
    'pct_ruim_muito_ruim'
] = (

    (
        notas_bloco['Ruim']
        +
        notas_bloco['Muito Ruim']
    )

    /

    notas_bloco[
        'total_notas'
    ]

    * 100

).round(2)


variacao_bloco = (

    variacao_bloco

    .merge(

        notas_bloco[
            [
                'bloco',
                'total_notas',
                'pct_ruim_muito_ruim'
            ]
        ],

        on='bloco',

        how='left'

    )

    .sort_values(

        'pct_grupos_com_variacao',

        ascending=False

    )

)


# ============================================================
# 7 - VARIAÇÃO POR AVALIADOR
# ============================================================

variacao_avaliador = (

    variacao_detalhada

    .groupby(
        'avaliador',
        as_index=False
    )

    .agg(

        qtd_escolas=(
            'escola',
            'nunique'
        ),

        qtd_grupos_comparados=(
            'houve_variacao',
            'size'
        ),

        qtd_grupos_com_variacao=(
            'houve_variacao',
            lambda x:
            (x == 'Sim').sum()
        ),

        qtd_avaliacoes=(
            'qtd_avaliacoes',
            'sum'
        )

    )

)


variacao_avaliador[
    'pct_grupos_com_variacao'
] = (

    variacao_avaliador[
        'qtd_grupos_com_variacao'
    ]

    /

    variacao_avaliador[
        'qtd_grupos_comparados'
    ]

    * 100

).round(2)


variacao_avaliador = (

    variacao_avaliador

    .sort_values(

        'pct_grupos_com_variacao',

        ascending=False

    )

)


# ============================================================
# 8 - VARIAÇÃO POR ESCOLA
# ============================================================

variacao_escola = (

    variacao_detalhada

    .groupby(
        'escola',
        as_index=False
    )

    .agg(

        qtd_avaliadores=(
            'avaliador',
            'nunique'
        ),

        qtd_grupos_comparados=(
            'houve_variacao',
            'size'
        ),

        qtd_grupos_com_variacao=(
            'houve_variacao',
            lambda x:
            (x == 'Sim').sum()
        ),

        qtd_avaliacoes=(
            'qtd_avaliacoes',
            'sum'
        )

    )

)


variacao_escola[
    'pct_grupos_com_variacao'
] = (

    variacao_escola[
        'qtd_grupos_com_variacao'
    ]

    /

    variacao_escola[
        'qtd_grupos_comparados'
    ]

    * 100

).round(2)


variacao_escola = (

    variacao_escola

    .sort_values(

        [
            'qtd_grupos_com_variacao',
            'pct_grupos_com_variacao'
        ],

        ascending=False

    )

)


# ============================================================
# 9 - VARIAÇÃO POR ESCOLA + AVALIADOR
# ============================================================

variacao_escola_avaliador = (

    variacao_detalhada

    .groupby(

        [
            'escola',
            'avaliador'
        ],

        as_index=False

    )

    .agg(

        qtd_grupos_comparados=(
            'houve_variacao',
            'size'
        ),

        qtd_grupos_com_variacao=(
            'houve_variacao',
            lambda x:
            (x == 'Sim').sum()
        ),

        qtd_avaliacoes=(
            'qtd_avaliacoes',
            'sum'
        )

    )

)


variacao_escola_avaliador[
    'pct_grupos_com_variacao'
] = (

    variacao_escola_avaliador[
        'qtd_grupos_com_variacao'
    ]

    /

    variacao_escola_avaliador[
        'qtd_grupos_comparados'
    ]

    * 100

).round(2)


variacao_escola_avaliador = (

    variacao_escola_avaliador

    .sort_values(

        [
            'pct_grupos_com_variacao',
            'qtd_grupos_comparados'
        ],

        ascending=False

    )

)


# ============================================================
# 10 - ANÁLISE MENSAL
# ============================================================

analise_mensal = (

    long_df

    .groupby(
        [
            'mes',
            'nota'
        ]
    )

    .size()

    .unstack(
        fill_value=0
    )

    .reindex(
        columns=NOTAS,
        fill_value=0
    )

    .reset_index()

)


analise_mensal[
    'total'
] = (

    analise_mensal[
        NOTAS
    ]
    .sum(axis=1)

)


for nota in NOTAS:

    coluna = (
        'pct_'
        +
        nota.lower()
        .replace(' ', '_')
    )

    analise_mensal[coluna] = (

        analise_mensal[nota]

        /

        analise_mensal[
            'total'
        ]

        * 100

    ).round(2)


analise_mensal[
    'pct_ruim_muito_ruim'
] = (

    (
        analise_mensal['Ruim']
        +
        analise_mensal['Muito Ruim']
    )

    /

    analise_mensal[
        'total'
    ]

    * 100

).round(2)


# ============================================================
# 11 - MATRIZ DE TRANSIÇÃO
# ============================================================

matriz_transicao = pd.crosstab(

    transicoes[
        'nota_anterior'
    ],

    transicoes[
        'nota'
    ]

)


matriz_transicao = matriz_transicao.reindex(

    index=NOTAS,

    columns=NOTAS,

    fill_value=0

)


# ============================================================
# 12 - QUALIDADE DOS DADOS
# ============================================================

qualidade_dados = []


# ------------------------------------------------------------
# Cobertura de cada critério
# ------------------------------------------------------------

for criterio in criterios_existentes:

    total_preenchido = (

        df[criterio]
        .isin(NOTAS)
        .sum()

    )

    qualidade_dados.append({

        'tipo':
            'Cobertura',

        'item':
            criterio,

        'valor':
            int(total_preenchido),

        'percentual':
            round(
                total_preenchido
                /
                len(df)
                *
                100,
                2
            )

    })


# ------------------------------------------------------------
# Campos idênticos
# ------------------------------------------------------------

for i, coluna1 in enumerate(
    criterios_existentes
):

    for coluna2 in (
        criterios_existentes[
            i + 1:
        ]
    ):

        s1 = (
            df[coluna1]
            .astype('string')
        )

        s2 = (
            df[coluna2]
            .astype('string')
        )

        if s1.equals(s2):

            qualidade_dados.append({

                'tipo':
                    'Colunas idênticas',

                'item':
                    f'{coluna1} = {coluna2}',

                'valor':
                    len(df),

                'percentual':
                    100.0

            })


qualidade_dados = pd.DataFrame(
    qualidade_dados
)


# ============================================================
# 13 - GRÁFICO - DISTRIBUIÇÃO DAS NOTAS
# ============================================================

plt.figure(
    figsize=(9, 5)
)

plt.bar(

    distribuicao_notas[
        'nota'
    ],

    distribuicao_notas[
        'percentual'
    ]

)

plt.title(
    'Distribuição geral das notas'
)

plt.ylabel(
    'Percentual (%)'
)

plt.xlabel(
    'Nota'
)

plt.xticks(
    rotation=20
)

plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '01_distribuicao_notas.png',

    dpi=200

)

plt.close()


# ============================================================
# 14 - GRÁFICO - VARIAÇÃO POR CRITÉRIO
# ============================================================

grafico = (

    variacao_criterio

    .sort_values(
        'pct_grupos_com_variacao'
    )

)


plt.figure(
    figsize=(10, 10)
)

plt.barh(

    grafico[
        'criterio'
    ],

    grafico[
        'pct_grupos_com_variacao'
    ]

)

plt.xlabel(
    '% de grupos com variação'
)

plt.ylabel(
    'Critério'
)

plt.title(
    'Variação das notas por critério'
)

plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '02_variacao_por_criterio.png',

    dpi=200

)

plt.close()


# ============================================================
# 15 - GRÁFICO - RUIM + MUITO RUIM
# ============================================================

grafico = (

    variacao_criterio

    .sort_values(
        'pct_ruim_muito_ruim'
    )

)


plt.figure(
    figsize=(10, 10)
)

plt.barh(

    grafico[
        'criterio'
    ],

    grafico[
        'pct_ruim_muito_ruim'
    ]

)

plt.xlabel(
    '% Ruim + Muito Ruim'
)

plt.ylabel(
    'Critério'
)

plt.title(
    'Participação de Ruim e Muito Ruim por critério'
)

plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '03_ruim_muito_ruim_criterio.png',

    dpi=200

)

plt.close()


# ============================================================
# 16 - GRÁFICO - AVALIADORES
#
# Para evitar um gráfico ilegível com todos os avaliadores,
# são mostrados os 20 com maior quantidade de grupos
# comparáveis.
#
# Isso NÃO exclui avaliadores do Excel.
# ============================================================

grafico_avaliador = (

    variacao_avaliador

    .nlargest(
        20,
        'qtd_grupos_comparados'
    )

    .sort_values(
        'pct_grupos_com_variacao'
    )

)


plt.figure(
    figsize=(11, 8)
)

plt.barh(

    grafico_avaliador[
        'avaliador'
    ],

    grafico_avaliador[
        'pct_grupos_com_variacao'
    ]

)

plt.xlabel(
    '% de grupos com variação'
)

plt.ylabel(
    'Avaliador'
)

plt.title(
    'Variação dos 20 avaliadores com maior volume comparável'
)

plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '04_variacao_avaliadores.png',

    dpi=200

)

plt.close()


# ============================================================
# 17 - GRÁFICO - EVOLUÇÃO MENSAL
# ============================================================

plt.figure(
    figsize=(10, 6)
)


for nota in NOTAS:

    coluna = (
        'pct_'
        +
        nota.lower()
        .replace(' ', '_')
    )

    plt.plot(

        analise_mensal[
            'mes'
        ],

        analise_mensal[
            coluna
        ],

        marker='o',

        label=nota

    )


plt.title(
    'Evolução mensal das classificações'
)

plt.xlabel(
    'Mês'
)

plt.ylabel(
    'Percentual (%)'
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '05_evolucao_mensal.png',

    dpi=200

)

plt.close()


# ============================================================
# 18 - MATRIZ DE TRANSIÇÃO
# ============================================================

fig, ax = plt.subplots(
    figsize=(8, 7)
)


imagem = ax.imshow(
    matriz_transicao.values
)


ax.set_xticks(
    range(len(NOTAS))
)

ax.set_yticks(
    range(len(NOTAS))
)

ax.set_xticklabels(
    NOTAS,
    rotation=30,
    ha='right'
)

ax.set_yticklabels(
    NOTAS
)


ax.set_xlabel(
    'Nota atual'
)

ax.set_ylabel(
    'Nota anterior'
)

ax.set_title(
    'Matriz de transição das avaliações'
)


for i in range(
    len(NOTAS)
):

    for j in range(
        len(NOTAS)
    ):

        ax.text(

            j,
            i,

            f'{matriz_transicao.iloc[i, j]:,}',

            ha='center',
            va='center'

        )


plt.tight_layout()

plt.savefig(

    PASTA_GRAFICOS /
    '06_matriz_transicao.png',

    dpi=200

)

plt.close()


# ============================================================
# 19 - EXPORTAÇÃO PARA EXCEL
# ============================================================

print('\n6. Exportando Excel gerencial...')


with pd.ExcelWriter(

    ARQUIVO_RESULTADO,

    engine='openpyxl'

) as writer:


    resumo_geral.to_excel(

        writer,
        sheet_name='00_resumo_geral',
        index=False

    )


    distribuicao_notas.to_excel(

        writer,
        sheet_name='01_distribuicao_notas',
        index=False

    )


    variacao_criterio.to_excel(

        writer,
        sheet_name='02_criterios',
        index=False

    )


    variacao_bloco.to_excel(

        writer,
        sheet_name='03_blocos',
        index=False

    )


    variacao_avaliador.to_excel(

        writer,
        sheet_name='04_avaliadores',
        index=False

    )


    variacao_escola.to_excel(

        writer,
        sheet_name='05_escolas',
        index=False

    )


    variacao_escola_avaliador.to_excel(

        writer,
        sheet_name='06_escola_avaliador',
        index=False

    )


    variacao_detalhada.to_excel(

        writer,
        sheet_name='07_variacao_detalhada',
        index=False

    )


    resumo_transicoes.to_excel(

        writer,
        sheet_name='08_transicoes',
        index=False

    )


    estabilidade_visitas.to_excel(

        writer,
        sheet_name='09_estabilidade',
        index=False

    )


    matriz_transicao.to_excel(

        writer,
        sheet_name='10_matriz_transicao'

    )


    analise_mensal.to_excel(

        writer,
        sheet_name='11_mensal',
        index=False

    )


    qualidade_dados.to_excel(

        writer,
        sheet_name='12_qualidade_dados',
        index=False

    )


# ============================================================
# 20 - FORMATAÇÃO DO EXCEL
# ============================================================

wb = load_workbook(
    ARQUIVO_RESULTADO
)


for ws in wb.worksheets:

    # congela cabeçalho
    ws.freeze_panes = 'A2'

    # autofiltro
    ws.auto_filter.ref = (
        ws.dimensions
    )

    # cabeçalho
    for cell in ws[1]:

        cell.font = Font(
            bold=True
        )

        cell.alignment = Alignment(
            horizontal='center',
            vertical='center'
        )

    # largura automática
    for coluna in ws.columns:

        tamanho_maximo = 0

        letra = get_column_letter(
            coluna[0].column
        )

        for cell in coluna:

            try:

                tamanho = len(
                    str(cell.value)
                )

                if tamanho > tamanho_maximo:
                    tamanho_maximo = tamanho

            except Exception:
                pass

        ws.column_dimensions[
            letra
        ].width = min(
            tamanho_maximo + 2,
            40
        )


wb.save(
    ARQUIVO_RESULTADO
)


# ============================================================
# 21 - PRINCIPAIS NÚMEROS NO TERMINAL
# ============================================================

print('\n')
print('=' * 100)
print('RESUMO GERENCIAL')
print('=' * 100)


print(
    f'\nRegistros da base: '
    f'{len(df):,}'
)


print(
    f'Escolas: '
    f'{df["escola"].nunique():,}'
)


print(
    f'Avaliadores: '
    f'{df["avaliador"].nunique():,}'
)


print(
    f'Avaliações categóricas: '
    f'{len(long_df):,}'
)


print(
    f'Grupos comparáveis: '
    f'{len(variacao_detalhada):,}'
)


qtd_variacao = (

    variacao_detalhada[
        'houve_variacao'
    ]
    .eq('Sim')
    .sum()

)


pct_variacao = (

    qtd_variacao

    /

    len(
        variacao_detalhada
    )

    * 100

)


print(
    f'Grupos com variação: '
    f'{qtd_variacao:,} '
    f'({pct_variacao:.2f}%)'
)


# ============================================================
# MUDANÇAS ENTRE VISITAS
# ============================================================

qtd_transicoes = len(
    transicoes
)


qtd_mudancas = (

    transicoes[
        'mudou_nota'
    ]
    .eq('Sim')
    .sum()

)


pct_mudancas = (

    qtd_mudancas

    /

    qtd_transicoes

    * 100

)


print(
    f'\nComparações entre visitas consecutivas: '
    f'{qtd_transicoes:,}'
)


print(
    f'Mudanças de categoria: '
    f'{qtd_mudancas:,} '
    f'({pct_mudancas:.2f}%)'
)


print(
    f'Sem mudança de categoria: '
    f'{qtd_transicoes - qtd_mudancas:,} '
    f'({100 - pct_mudancas:.2f}%)'
)


# ============================================================
# CRITÉRIOS COM MAIOR VARIAÇÃO
# ============================================================

print('\n')
print('=' * 100)
print('CRITÉRIOS COM MAIOR VARIAÇÃO')
print('=' * 100)


print(

    variacao_criterio[
        [
            'criterio',
            'qtd_grupos_comparados',
            'qtd_grupos_com_variacao',
            'pct_grupos_com_variacao',
            'pct_ruim_muito_ruim'
        ]
    ]

    .head(15)

    .to_string(
        index=False
    )

)


# ============================================================
# CONCLUSÃO
# ============================================================

print('\n')
print('=' * 100)
print('PROCESSAMENTO FINALIZADO')
print('=' * 100)


print(
    f'\nExcel gerado:\n'
    f'{ARQUIVO_RESULTADO}'
)


print(
    f'\nGráficos gerados em:\n'
    f'{PASTA_GRAFICOS}'
)
```
